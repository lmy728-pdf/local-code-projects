选用 environment2 内核。运行下方单元，从正式检查点复现 D3 样本并生成两个方法的区间 CSV 和覆盖率图。

```python
from pathlib import Path
import json
import os
import subprocess
from IPython.display import clear_output

code_dir = Path.cwd()
script = next((folder / "replay_d3_sandwich.py" for folder in (
    code_dir,
    code_dir / "D3实验_代码",
    code_dir / "D3_Sandwich补算" / "D3实验_代码",
    Path.home() / "Desktop" / "毕设论文实验(这个是正式版，千万别搞错)" /
        "Rank_DML_Scenario_D" / "D3_Sandwich补算" / "D3实验_代码",
) if (folder / "replay_d3_sandwich.py").is_file()), None)
if script is None:
    raise FileNotFoundError("请从补算文件夹打开本 Notebook。")
versions = json.loads((script.parents[2] / "environment_versions.json").read_text(encoding="utf-8"))
with subprocess.Popen(
    [versions["verified_interpreter"], "-B", str(script), "--workers", "12"],
    cwd=script.parent, text=True, encoding="utf-8", errors="replace", stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT, env=os.environ | {"PYTHONUTF8": "1"},
    creationflags=getattr(subprocess, "CREATE_NO_WINDOW", 0),
) as process:
    for line in process.stdout:
        if line.startswith("D3 covariance"):
            clear_output(wait=True)
        print(line.strip())
    if process.wait():
        raise RuntimeError("补算退出，请查看单元输出。")
print(script.parents[1])
```
